```python
import tensorflow as tf
import os

# 1. Load model Neural Network yang sudah dilatih sebelumnya
model_path = '../models/runform_model.h5'
model = tf.keras.models.load_model(model_path)

# 2. Inisialisasi Converter
converter = tf.lite.TFLiteConverter.from_keras_model(model)

# 3. Optimasi (Dynamic Range Quantization)
# Ini akan mengompres ukuran model agar ringan di HP tanpa banyak mengurangi akurasi
converter.optimizations = [tf.lite.Optimize.DEFAULT]

tflite_model = converter.convert()

# 4. Simpan hasil konversi
tflite_output_path = '../models/runform_model.tflite'
with open(tflite_output_path, 'wb') as f:
    f.write(tflite_model)

print(f"Model berhasil dikonversi ke TFLite: {tflite_output_path}")
```

```python
# Cek ukuran file
h5_size = os.path.getsize(model_path) / 1024
tflite_size = os.path.getsize(tflite_output_path) / 1024

print(f"Ukuran Model Keras (.h5)  : {h5_size:.2f} KB")
print(f"Ukuran Model TFLite (.tflite): {tflite_size:.2f} KB")
print(f"Efisiensi Ruang: {((h5_size - tflite_size) / h5_size) * 100:.2f}%")
```
